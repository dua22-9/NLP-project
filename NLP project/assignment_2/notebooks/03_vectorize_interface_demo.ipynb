This notebook demonstrates the unified `get_document_vector()` interface designed for Phase 3 downstream machine learning models.


```python
import os
import sys
import numpy as np

# Add src to path
sys.path.insert(0, os.path.abspath("../src"))
from vectorize import get_document_vector

sample_sentence = "apple announce new battery upgrade for iphone"
print(f"Sample Input: "{sample_sentence}"\n")

for model in ["bow", "tfidf", "word2vec_cbow", "word2vec_skipgram", "fasttext"]:
    vec = get_document_vector(sample_sentence, model_type=model, pooling="mean", models_dir="../models")
    print(f"{model:<20s} | Shape: {str(vec.shape):<10s} | Non-Zero: {np.count_nonzero(vec):<6d} | Sample: {vec[:3]}")
```

### Out-of-Vocabulary Resilience Test
Comparing Word2Vec vs fastText on unseen input words.


```python
oov_text = "xyzfakeoovword qwertyunknownword zzzcompletelybogustoken"

vec_sg = get_document_vector(oov_text, model_type="word2vec_skipgram", models_dir="../models")
vec_ft = get_document_vector(oov_text, model_type="fasttext", models_dir="../models")

print(f"Word2Vec Skip-gram Non-Zero Entries: {np.count_nonzero(vec_sg)} (Zero vector fallback)")
print(f"fastText Sub-word Non-Zero Entries:  {np.count_nonzero(vec_ft)} (Full 100d Vector constructed!)")
```
