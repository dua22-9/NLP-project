## 1. Load the Raw Social Media Corpus

## 2. Execute Corpus Statistics & Exploration
Computes word length statistics, label distribution, vocabulary size, and Zipf's Law distribution.

```python
stats = run_exploration()
```

## 3. Visualizing Target Distribution and Word Lengths

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
# Class distribution
df['sentiment'].value_counts().plot(kind='bar', ax=axes[0], color=['#2ca02c', '#d62728', '#1f77b4'], edgecolor='black')
axes[0].set_title('Target Class Distribution')
axes[0].set_xlabel('Sentiment')
axes[0].set_ylabel('Document Count')

# Length distribution
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
axes[1].hist(df['word_count'], bins=15, color='#4a90e2', edgecolor='black')
axes[1].set_title('Document Word Length Distribution')
axes[1].set_xlabel('Word Count')
axes[1].set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## 4. Inspection of Real Messy Text Examples
Social media text features HTML artifacts, URLs, handles, punctuation floods, and mixed casing.

```python
for eg in stats['examples']:
    print(f'=== Example #{eg["index"]} [{eg["label"]}] ===')
    print('Raw text: ', repr(eg['raw']))
    print('Anomalies:', eg['issues'])
    print()
```
